```python
class Musica:
    musicas = []
    # __init__ Método construtor: Cria parâmetros obrigatórios para a criação do ojbeto à partir da classe
    def __init__(self, nome, artista, duracao):
        self._nome = nome.title()
        self._artista = artista 
        self.duracao = duracao 
        self._ativa = False

        Musica.musicas.append(self)
    # __str__ método de exibição textual da classe 
    # Toda vez que eu der o comando print(objeto instanciado), o texto retornado será o que estiver no return aqui
    def __str__(self):
        return f'{self._nome.ljust(25)} | {self.artista.ljust(25)} | {self.duracao}'

    # @classmethod + cls é uma convenção para estruturar um método de classe,indicando que o método é dedicado a classe, e não 
    # à instancia criada a partir dela. 
    @classmethod
    def exibir_musicas(cls):
        print(f'{"Musica".ljust(25)} | {"Artista".ljust(25)} | {"Duração".ljust(25)} | {"Status"}')
        for musica in cls.musicas:
            print(f"{musica._nome.ljust(25)} | {musica._artista.ljust(25)} | {musica.duracao.ljust(25)} | {musica.ativa}") 

    # @property + underscore no nome da variável
    # Permite o encapsulamento de uma variável, camuflando um método como variável. Se o usuário chamar obj.ativa, ele receberá
    # o retorno com o valor de acordo com o resultado do bloco if, porém, não poderá alterar o valor, uma vez que ele está sendo 
    # pelo método "ativa"
    # O underscore é uma sinalização visual para quem for utilizar a classe de que aquela variável não deve ser editada.
    @property
    def ativa(self):
        return "Ativa" if self._ativa else "Inativa"

    # Método de instância/objeto: Vai atuar mudando o status exclusivamente no objeto instanciado a partir da classe
    def alternar_status(self):
        self._ativa = not self._ativa 

musica1 = Musica("TESTE1", "NOME1", "5")
musica2 = Musica("TESTE2", "NOME2", "10")
musica3 = Musica("TESTE3", "NOME3", "2") 

musica3.alternar_status()

Musica.exibir_musicas()
```
