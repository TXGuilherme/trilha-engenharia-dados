```python
class Musica:
    musicas = []
    # __init__ Método construtor: Cria parâmetros obrigatórios para a criação do ojbeto à partir da classe
    def __init__(self, nome, artista, duracao):
        self.nome = nome 
        self.artista = artista 
        self.duracao = duracao 

        Musica.musicas.append(self)
    # __str__ método de exibição textual da classe 
    # Toda vez que eu der o comando print(objeto instanciado), o texto retornado será o que estiver no return aqui
    def __str__(self):
        return f'{self.nome} | {self.artista} | {self.duracao}'
    
    def exibir_musicas():
        for musica in Musica.musicas:
            print(f"Nome da música: {musica.nome} Artista: {musica.artista} Duração: {musica.duracao} minutos\n") 
            

musica1 = Musica("TESTE1", "NOME1", 5)
musica2 = Musica("TESTE2", "NOME2", 10)
musica3 = Musica("TESTE3", "NOME3", 2) 

Musica.exibir_musicas()
```
